```python
import numpy as np
# ============================================================
# PART 1: ARRAY CREATION METHODS
# ============================================================
print("========== ARRAY CREATION METHODS ==========")

# 1. np.array()
# Creates an array from an existing list or tuple
marks = np.array([60, 70, 80, 90, 100])
print("\n1. np.array()")
print("Marks:", marks)
# 2. np.zeros()
# Creates an array filled with zeros
zero_array = np.zeros(5)
print("\n2. np.zeros()")
print(zero_array)
# 3. np.ones()
# Creates an array filled with ones
one_array = np.ones(5)
print("\n3. np.ones()")
print(one_array)
# 4. np.full()
# Creates an array filled with the same specified value
full_array = np.full(5, 50)
print("\n4. np.full()")
print(full_array)
# 5. np.empty()
# Creates an array without initializing values
empty_array = np.empty(5)
print("\n5. np.empty()")
print(empty_array)
# 6. np.arange()
# Creates values using start, stop and step
range_array = np.arange(10, 60, 10)
print("\n6. np.arange()")
print(range_array)
# 7. np.linspace()
# Creates a fixed number of equally spaced values
linear_array = np.linspace(0, 100, 5)
print("\n7. np.linspace()")
print(linear_array)
# 8. np.eye()
# Creates a matrix having 1 on the required diagonal
eye_matrix = np.eye(3)
print("\n8. np.eye()")
print(eye_matrix)
# 9. np.identity()
# Creates a square identity matrix
identity_matrix = np.identity(3)
print("\n9. np.identity()")
print(identity_matrix)

# ============================================================
# PART 2: STATISTICAL FUNCTIONS
# ============================================================

print("\n\n========== STATISTICAL METHODS ==========")

data = np.array([60, 70, 80, 90, 100])

print("\nOriginal Data:")
print(data)

# 10. np.mean()
# Calculates arithmetic mean
print("\n10. Mean:")
print(np.mean(data))
# 11. np.average()
# Calculates average
print("\n11. Average:")
print(np.average(data))
# 12. Weighted Average
weights = np.array([1, 1, 2, 2, 4])
print("\n12. Weighted Average:")
print(np.average(data, weights=weights))
# 13. np.median()
# Gives middle value
print("\n13. Median:")
print(np.median(data))


# 14. np.min()
# Gives minimum value
print("\n14. Minimum:")
print(np.min(data))
# 15. np.max()
# Gives maximum value
print("\n15. Maximum:")
print(np.max(data))
# 16. np.ptp()
# Range = Maximum - Minimum
print("\n16. Range:")
print(np.ptp(data))
# 17. np.argmin()
# Gives index of minimum value
print("\n17. Index of Minimum Value:")
print(np.argmin(data))
# 18. np.argmax()
# Gives index of maximum value
print("\n18. Index of Maximum Value:")
print(np.argmax(data))
# 19. np.std()
# Calculates standard deviation
print("\n19. Standard Deviation:")
print(np.std(data))
# 20. np.var()
# Calculates variance
print("\n20. Variance:")
print(np.var(data))
# 21. np.percentile()
#position=(n-1)*p/100
print("\n21. 25th Percentile:")
print(np.percentile(data, 25))
print("50th Percentile:")
print(np.percentile(data, 50))
print("75th Percentile:")
print(np.percentile(data, 75))
# 22. np.quantile()
print("\n22. First Quantile:")
print(np.quantile(data, 0.25))
print("Second Quantile:")
print(np.quantile(data, 0.50))
print("Third Quantile:")
print(np.quantile(data, 0.75))
# ============================================================
# PART 3: CORRELATION AND COVARIANCE
# ============================================================

print("\n\n========== RELATIONSHIP ANALYSIS ==========")

study_hours = np.array([1, 2, 3, 4, 5])

student_marks = np.array([50, 60, 70, 80, 90])
print("\nStudy Hours:")
print(study_hours)
print("\nStudent Marks:")
print(student_marks)
# 23. np.corrcoef()
# Finds correlation between two variables
print("\n23. Correlation Matrix:")
print(np.corrcoef(study_hours, student_marks))

print("\nCorrelation Value:")
print(np.corrcoef(study_hours, student_marks)[0, 1])
# 24. np.cov()
# Finds covariance between two variables
print("\n24. Covariance:")
print(np.cov(study_hours, student_marks))

# 25. np.correlate()
# Cross-correlation between sequences

print("\n25. Cross Correlation:")
print(np.correlate(study_hours, student_marks))
# ============================================================
# PART 4: STATISTICAL METHODS WITH MISSING VALUES
# ============================================================
print("\n\n========== STATISTICS WITH MISSING VALUES ==========")

missing_marks = np.array([60, 70, np.nan, 90, 100])

print("\nData:")
print(missing_marks)


# 26. np.nanmean()

print("\n26. Mean Ignoring NaN:")
print(np.nanmean(missing_marks))


# 27. np.nanmedian()

print("\n27. Median Ignoring NaN:")
print(np.nanmedian(missing_marks))


# 28. np.nanstd()

print("\n28. Standard Deviation Ignoring NaN:")
print(np.nanstd(missing_marks))


# 29. np.nanvar()

print("\n29. Variance Ignoring NaN:")
print(np.nanvar(missing_marks))


# 30. np.nanmin()

print("\n30. Minimum Ignoring NaN:")
print(np.nanmin(missing_marks))


# 31. np.nanmax()

print("\n31. Maximum Ignoring NaN:")
print(np.nanmax(missing_marks))


# 32. np.nanpercentile()

print("\n32. 75th Percentile Ignoring NaN:")
print(np.nanpercentile(missing_marks, 75))


# 33. np.nanquantile()

print("\n33. 0.75 Quantile Ignoring NaN:")
print(np.nanquantile(missing_marks, 0.75))

```
